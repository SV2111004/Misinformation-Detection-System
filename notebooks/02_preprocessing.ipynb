```python
import pandas as pd
import re
```

```python
columns = [
    "id",
    "label",
    "statement",
    "subject",
    "speaker",
    "speaker_job",
    "state",
    "party",
    "barely_true",
    "false",
    "half_true",
    "mostly_true",
    "pants_fire",
    "context"
]
```

```python
train = pd.read_csv(
    "../dataset/liar/train.tsv",
    sep="\t",
    header=None,
    names=columns
)

valid = pd.read_csv(
    "../dataset/liar/valid.tsv",
    sep="\t",
    header=None,
    names=columns
)

test = pd.read_csv(
    "../dataset/liar/test.tsv",
    sep="\t",
    header=None,
    names=columns
)
```

```python
train_ml = train[["statement", "label"]].copy()
valid_ml = valid[["statement", "label"]].copy()
test_ml = test[["statement", "label"]].copy()
```

```python
print(train_ml.shape)
print(valid_ml.shape)
print(test_ml.shape)
```

```python
train_ml.head()
```

```python
train_ml = train_ml.dropna(subset=["statement", "label"])
valid_ml = valid_ml.dropna(subset=["statement", "label"])
test_ml = test_ml.dropna(subset=["statement", "label"])
print(train_ml.isnull().sum())
print(valid_ml.isnull().sum())
print(test_ml.isnull().sum())
```

```python
train_ml = train_ml.drop_duplicates()
valid_ml = valid_ml.drop_duplicates()
test_ml = test_ml.drop_duplicates()
```

```python
def clean_text(text):
    text = str(text)
    text = text.strip()
    text = re.sub(r"\s+", " ", text)
    return text

train_ml["statement"] = train_ml["statement"].apply(clean_text)
valid_ml["statement"] = valid_ml["statement"].apply(clean_text)
test_ml["statement"] = test_ml["statement"].apply(clean_text)
```

```python
print((train_ml["statement"].str.len() == 0).sum())
print((valid_ml["statement"].str.len() == 0).sum())
print((test_ml["statement"].str.len() == 0).sum())
```

```python
print(train_ml["label"].unique())
```

```python
print(train_ml["label"].value_counts())
```

```python
X_train = train_ml["statement"]
y_train = train_ml["label"]

X_valid = valid_ml["statement"]
y_valid = valid_ml["label"]

X_test = test_ml["statement"]
y_test = test_ml["label"]
```

```python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_valid:", X_valid.shape)
print("y_valid:", y_valid.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
print("Training samples:", len(X_train))
print("Validation samples:", len(X_valid))
print("Test samples:", len(X_test))

print("\nLabels:")
print(y_train.value_counts())

print("\nSample:")
print(X_train.iloc[0])
print("Label:", y_train.iloc[0])
```
